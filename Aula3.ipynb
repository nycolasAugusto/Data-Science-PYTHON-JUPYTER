```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


dados =['mobile', 'tablet', 'laptop', 'desktop','desktop','mobile']

df = pd.DataFrame({'dispositivo': dados}) 

sns.countplot(
    data=df,
    x='dispositivo'
)
plt.show()
#Gera um grafico conforme a quantidade de cada dispositivo na lista de dados. nao trabalha so com NUMEROS
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd


regioes = ['Norte', 'Sul', 'Norte', 'Leste', 'Sul', 'Leste', 'Norte']
faturamentos = [1500, 2300, 1200, 3100, 2800, 1900, 2000]

df = pd.DataFrame({
    'regiao': regioes,
    'faturamento': faturamentos
}) 

# Eixo X = Categoria | Eixo Y = Número[cite: 2]
sns.barplot(
    data=df,
    x='regiao',
    y='faturamento'
)
plt.show()

# Gera um gráfico de barras relacionando uma categoria (texto) com um valor numérico (faturamento).
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

datas_compra = ['01/05', '02/05', '03/05', '04/05', '05/05']
faturamentos = [1200, 1500, 900, 2300, 1750]

df = pd.DataFrame({
    'data_compra': datas_compra,
    'faturamento': faturamentos
}) 

# Eixo X = Categoria | Eixo Y = Número[cite: 2]
sns.lineplot(
    data=df,
    x='data_compra',
    y='faturamento'
)
plt.show()

# Gera um gráfico de barras relacionando uma categoria (texto) com um valor numérico (faturamento).
```
