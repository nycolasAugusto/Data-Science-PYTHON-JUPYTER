```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd


regioes = ['Norte', 'Sul', 'Norte', 'Leste', 'Sul', 'Leste', 'Norte']
faturamentos = [1500, 2300, 1200, 3100, 2800, 1900, 2000]

df = pd.DataFrame({
    'regiao': regioes,
    'faturamento': faturamentos
}) 

# Eixo X = Categoria | Eixo Y = Número[cite: 2]
sns.barplot(
    data=df,
    x='regiao',
    y='faturamento'
)
plt.show()

# Gera um gráfico de barras relacionando uma categoria (texto) com um valor numérico (faturamento).
```

```python
from scipy import stats

# Dados de exemplo para o faturamento de duas versões (A e B)
array_faturamento_A = [100, 120, 110, 130, 115]
array_faturamento_B = [140, 150, 145, 160, 155]

# 1. Invocando o Juiz (A máquina faz o cálculo interno)[cite: 3]
t_stat, p_valor = stats.ttest_ind(
    array_faturamento_A,
    array_faturamento_B
)

# 2. Imprimindo formatado com 4 casas decimais[cite: 3]
print(f"O P-Valor calculado foi: {p_valor:.4f}")

# A Regra dos 5% de Alfa[cite: 4]
if p_valor < 0.05:
    print("Sucesso! A diferença é estatisticamente real (H1).")
    print("Ação: Fazer deploy da Versão B para 100% da base.")
else:
    print("Fracasso! É puro ruído e sorte (H0).")[cite: 4]
    print("Ação: Manter o design da Versão A e cancelar bônus.")[cite: 4]

# O teste t compara duas amostras para ver se a diferença entre elas é real ou acaso. Se o p-valor for menor que 0.05, aceitamos que a mudança deu certo; caso contrário, foi apenas variação aleatória.
```
