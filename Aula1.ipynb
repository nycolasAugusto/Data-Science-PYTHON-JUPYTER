```python
# MEDIA 
import pandas as pd

dados = [4000, 5000, 6000, 7000, 8000]

df = pd.DataFrame({'salarios': dados}) 
#dataFrame cria estrutura de TABELA 
#DF é a tabela -- salarios é a coluna 

media = df['salarios'].mean()

print(f"A média dos salários é: {media}")
```

```python
import pandas as pd

dados = ['400', '500', '500', '500', '800','400','400']


df = pd.DataFrame({'erros': dados}) 

moda = df['erros'].mode()[1]

#.mode()[1] retonrna a moda -> e [] serve para pegar qual moda caso tenha mais de 1

print(f"A moda dos erros é: {moda}")
```

```python
import pandas as pd

dados = [2, 5, 8]


df = pd.DataFrame({'horas': dados}) 

variancia = df['horas'].var()
desvio = df['horas'].std()

print(f"A variância das horas é: {variancia}")
print(f"O desvio padrão das horas é: {desvio}")
```
