## 1. Manipulação e Limpeza (O Preparo)

**Enunciado pede:** *"Isole em uma variável (tempo_A) todo o 'Tempo_Tela' apenas da 'Versao' A."*
* **O que ele quer:** Filtrar o DataFrame para separar grupos de análise.
* **Função no Python:** `tempo_A = df[df['Versao'] == 'A']['Tempo_Tela']`

**Enunciado pede:** *"Adequação à LGPD"*, *"Remova os dados sensíveis"*, *"Proteja a identidade"*.
* **O que ele quer:** Apagar a coluna que identifica a pessoa (Email, CPF, Cartão).
* **Função no Python:** `df.drop(columns=['Nome_da_Coluna'])`

**Enunciado pede:** *"Limpe a base para não enburrecer a IA"*, *"Tire os valores distorcidos"*.
* **O que ele quer:** Filtrar e remover os Outliers (valores absurdos).
* **Função no Python:** `df_limpo = df[df['Gasto'] < valor_limite]`

---

## 2. Estatística Básica e Outliers (O Diagnóstico)

**Enunciado pede:** *"Qual o perfil mais comum?"*, *"O que mais aparece?"*.
* **O que ele quer:** Descobrir a Moda (mais usado para textos/categorias).
* **Função no Python:** `df['coluna'].mode()[0]`

**Enunciado pede:** *"Verifique se há outliers analisando a média e a mediana"*.
* **O que ele quer:** Que você calcule os dois. Se a média for **MUITO maior ou menor** que a mediana, é certeza que existem outliers puxando o valor. A mediana é imune a outliers.
* **Função no Python:** `df['coluna'].mean()` e `df['coluna'].median()`

---

## 3. Gráficos (A Tradução Visual)

**Enunciado pede:** *"Descubra visualmente se há outliers"*.
* **O que ele quer:** Gráfico de Caixa (Boxplot). Os pontos que ficam flutuando para fora das linhas do gráfico são os outliers.
* **Função no Python:** `sns.boxplot(y=df['coluna'])`

**Enunciado pede:** *"Compare o gasto médio entre os planos/regiões"*.
* **O que ele quer:** Gráfico de Barras. Perfeito para ver quem ganha de quem (Categoria x Número).
* **Função no Python:** `sns.barplot(x='Categoria', y='Numero', data=df)`

**Enunciado pede:** *"Verifique se X e Y formam um padrão"*, *"Veja se eles andam juntos no gráfico"*.
* **O que ele quer:** Gráfico de Dispersão (Scatterplot). Ele plota pontinhos para cruzar duas colunas numéricas e ver se formam uma linha ou nuvem.
* **Função no Python:** `sns.scatterplot(x='Coluna1', y='Coluna2', data=df)`

**Enunciado pede:** *"Mostre a tabela de correlações de forma visual"*.
* **O que ele quer:** Mapa de Calor (Heatmap). Transforma uma tabela de números chatos num quadro colorido onde cores fortes mostram relações fortes.
* **Função no Python:** `sns.heatmap(matriz_de_correlacao, annot=True)`

---

## 4. Teste de Hipóteses (O Teste A/B)

**Enunciado pede:** *"Verifique se a nova versão funcionou ou se foi obra do acaso"*, *"Qual versão é matematicamente melhor?"*.
* **O que ele quer:** Teste T de Student. Você cruza o grupo A com o grupo B para extrair o `p-valor`.
* **Função no Python:** `stats.ttest_ind(grupo_A, grupo_B)`
* **A Regra de Ouro (p-valor):** 
  * Se der **MENOR que 0.05**: A diferença é real, a versão nova funcionou (Rejeita H0).
  * Se der **MAIOR que 0.05**: Tudo igual, diferença é mero acaso/ruído estatístico (Aceita H0).

---

## 5. IA e Regressão Linear (A Bola de Cristal)

**Enunciado pede:** *"Calcule a força da relação entre X e Y"*.
* **O que ele quer:** Correlação. Vai gerar um número de -1 a 1. Próximo de 1 (crescem juntos), próximo de -1 (um sobe, outro desce), próximo de 0 (não tem nada a ver).
* **Função no Python:** `df.corr()`

**Enunciado pede:** *"Crie um motor de previsão"*, *"Estime o faturamento"*.
* **O que ele quer:** Treinar um modelo de Regressão Linear.
* **Função no Python:** `modelo = LinearRegression()`, depois `modelo.fit(X, y)`

**Enunciado pede:** *"Qual a qualidade/precisão do modelo?"*.
* **O que ele quer:** R2 Score. Uma nota de 0 a 100% (0 a 1) dizendo o quanto seu modelo consegue explicar da realidade.
* **Função no Python:** `r2_score(y_real, previsoes)`

**Enunciado pede:** *"Qual a margem de erro?"*, *"Crie um cenário otimista e pessimista"*.
* **O que ele quer:** RMSE (Raiz do Erro Quadrático Médio). É o valor médio que seu modelo erra, na moeda real (ex: erra por R$ 50 para mais ou para menos). Cenário otimista = Previsão + RMSE. Cenário pessimista = Previsão - RMSE.
* **Função no Python:** `root_mean_squared_error(y_real, previsoes)`
