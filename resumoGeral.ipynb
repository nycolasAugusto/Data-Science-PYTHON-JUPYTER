## 📋 Dicionário de Bibliotecas

| Biblioteca | Função / Método | O que faz / Para que serve |
| :--- | :--- | :--- |

| **Pandas (`pd`)** | `pd.DataFrame({...})` | Cria uma estrutura de dados tabular (linhas e colunas), semelhante a uma planilha. |
| | `df.head()` | Exibe as primeiras 5 linhas de um DataFrame para inspeção rápida. |
| | `df.drop(columns=[...])` | Remove colunas ou linhas específicas do DataFrame (essencial para LGPD). |
| | `df['coluna'].mode()[0]` | Encontra a **Moda**, o valor mais frequente em uma coluna categórica. |
| | `df['coluna'].mean()` | Calcula a **Média aritmética** dos valores numéricos de uma coluna. |
| | `df['coluna'].median()` | Calcula a **Mediana** (o valor central dos dados ordenados). Robusta contra outliers. |
| | **`df.corr()`** | Calcula a **correlação de Pearson** entre as colunas numéricas (mede se elas crescem ou caem juntas em linha reta). |
| | **`df.corr(method='spearman')`**| Calcula a correlação olhando para a **ordem/posição** dos dados, ideal para relações que não formam uma linha reta perfeita. |

| **NumPy (`np`)** | `np.random.seed(n)` | Fixa uma "semente" para garantir que os dados sintéticos gerados sejam sempre iguais. |
| | `np.random.randint(...)` | Gera números inteiros aleatórios dentro de um intervalo definido. |
| | `np.random.choice(...)` | Seleciona itens aleatoriamente de uma lista com base em probabilidades. |
| | `np.random.normal(...)` | Gera números seguindo uma distribuição normal com média e desvio padrão. |
| | `np.concatenate(...)` | Junta dois ou mais arrays em um único array. |
| | `np.random.shuffle(...)` | Embaralha aleatoriamente a ordem dos elementos de um array. |
| | `np.abs(...)` | Retorna o valor absoluto de um número. |

| **Seaborn (`sns`)** | `sns.set_theme(...)` | Define o estilo visual global dos gráficos. |
| | `sns.barplot(...)` | Cria um gráfico de barras para comparar valores médios entre diferentes categorias. |
| | `sns.boxplot(...)` | Cria um gráfico de caixa (boxplot) para visualizar distribuição e **identificar outliers**. |
| | **`sns.scatterplot(...)`**| Desenha um **gráfico de dispersão** cruzando duas colunas para ver se formam um padrão (`x='A'`, `y='B'`). |
| | **`sns.heatmap(...)`** | Transforma uma matriz em um **mapa de calor colorido**. O `annot=True` escreve os números dentro dos quadrados. |

| **Matplotlib (`plt`)**| `plt.show()` | Renderiza e exibe os gráficos na tela. |

| **SciPy (`stats`)** | `stats.ttest_ind(...)` | Realiza o **Teste T**, comparando duas médias independentes e retornando o `p-valor`. |

| **Scikit-Learn**| `LinearRegression()` | Instancia o algoritmo de Regressão Linear. |
| | `modelo.fit(X, y)` | Treina o modelo de IA encontrando os melhores coeficientes (pesos). |
| | `modelo.predict(X)` | Gera novas previsões com base em dados de entrada (`X`). |
| | `modelo.coef_` | Retorna o **peso (inclinação da reta)** aprendido pelo modelo. |
| | `modelo.intercept_` | Retorna o **viés (intercepto no eixo y)**. |
| | `r2_score(...)` | Calcula o **R²**, indicando a porcentagem da variância do alvo explicada pelo modelo (0 a 1). |
| | `root_mean_squared_error(...)`| Calcula a **RMSE**, a margem de erro média do modelo na mesma unidade original. |

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error, r2_score

sns.set_theme(style="whitegrid")

# --- GERANDO A BASE SINTÉTICA DA STREAMFLIX ---
np.random.seed(10)
n_usuarios = 800

idades = np.random.randint(14, 75, n_usuarios)
cartoes = [f"1234-5678-9012-{1000+i}" for i in range(n_usuarios)]
planos = np.random.choice(['Basico', 'Premium', 'Familia'], n_usuarios)
interfaces = np.random.choice(['Antiga', 'Nova'], n_usuarios)

# Simulando tempo de tela diferente para o Teste A/B
tempo_antiga = np.random.normal(45, 15, 400)
tempo_nova = np.random.normal(58, 15, 400)
tempo_assistido = np.concatenate([tempo_antiga, tempo_nova])
np.random.shuffle(tempo_assistido)

# Simulando gasto com base no tempo assistido (relação linear)
mensalidade = (tempo_assistido * 0.6) + 15 + np.random.normal(0, 8, n_usuarios)

# Criando outliers (fraudes ou erros de sistema)
mensalidade[12] = 1200
mensalidade[405] = 950

df = pd.DataFrame({
    'Idade': idades,
    'Cartao_Credito': cartoes,
    'Plano': planos,
    'Interface_App': interfaces,
    'Tempo_Assistido_Min': np.abs(tempo_assistido),
    'Mensalidade_Paga_R$': np.abs(mensalidade)
})

print("Base da StreamFlix gerada com sucesso!")
display(df.head())
```
