```python
import pandas as pd

dados = [4000, 5000, 6000, 7000, 8000]

df = pd.DataFrame({'latencia': dados}) 

resumo = df['latencia'].describe()
 #describe devolve tudo sobre o dt, como média, desvio padrão, quartis, etc.

print(resumo)


```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

dados = [22, 25, 28, 30, 35, 40, 45, 60]
df = pd.DataFrame({'idade': dados})

sns.boxplot(data=df, x='idade')

plt.show()

#Esse grafico gerado mostra mais como os quartis funcionam, ignorando os outliers como o 60 
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

dados = [22, 25, 28, 30, 35, 40, 45, 60,60]
df = pd.DataFrame({'transacoes': dados})

sns.histplot(data=df, x='transacoes', bins=10)

plt.show()

#Esse grafico gerado mostra mais como os quartis funcionam, ignorando os outliers como o 60 
```
